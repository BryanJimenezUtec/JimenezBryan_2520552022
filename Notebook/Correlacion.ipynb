```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/BryanJimenezUtec/JimenezBryan_2520552022/refs/heads/main/Data/clave_H_correlacion.csv')
```

```python
# IDENTIFICAR VARIABLES NUMÉRICAS
df_numerico = df.select_dtypes(include=['int64', 'float64', 'int32', 'float32'])

# Mostramos la lista de las columnas numéricas que el código logró identificar
print("--- Variables numéricas listas para correlación ---")
print(df_numerico.columns.tolist())

# Vemos un resumen rápido para confirmar que todo esté en orden
print("\n--- Primeras filas de los datos numéricos ---")
print(df_numerico.head())
```

```python
print("=== 1. CONTROL DE CALIDAD Y VALORES NULOS ===")
# Revisamos si faltan datos en alguna columna
print(df.isnull().sum())

print("\n=== 2. INFORMACIÓN GENERAL DEL DATASET ===")
# Vemos los tipos de datos y cuántas filas reales se cargaron
print(df.info())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_num = df.select_dtypes(include=[np.number])
# 2. CALCULAR LA MATRIZ DE CORRELACIÓN
matriz_corr = df_num.corr()

print("--- Matriz de Correlación Numérica ---")
print(matriz_corr)
```

```python
# 1. Configurar el tamaño ideal de la ventana del gráfico
plt.figure(figsize=(10, 8))

# 2. Crear el mapa de calor con Seaborn
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt=".4f", linewidths=0.5)

# 3. Añadir un título formal para tu informe
plt.title('Mapa de Calor - Matriz de Correlación de Clientes', fontsize=14, fontweight='bold', pad=15)

# 4. Ajustar el diseño para que las etiquetas de las variables no se corten en las esquinas
plt.tight_layout()

# 5. Mostrar el gráfico en la pantalla de Colab
plt.show()
```
