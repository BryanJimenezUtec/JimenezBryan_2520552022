```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/BryanJimenezUtec/JimenezBryan_2520552022/refs/heads/main/Data/clave_H_asociacion.csv')
```

```python
print("--- Primeras filas del dataset ---")
print(df.head())
```

```python
# Mostramos la estructura general (tipos de datos y columnas)
print("\n--- Estructura del dataset ---")
print(df.info())
```

```python
# Verificar la cantidad total de valores nulos por cada columna
print("Valores nulos:")
print(df.isnull().sum())
```

```python
# Verificar la cantidad total de filas exactamente duplicadas
print("\nFilas duplicadas:")
print(df.duplicated().sum())
```

```python
# Importar librerías necesarias
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#  Crear la "cesta" agrupando por transacción y por artículo
cesta = df.groupby(['transaccion_id', 'item'])['cantidad'].sum().unstack().fillna(0)

#  Convertir las cantidades a formato binario (0 y 1)

cesta_binaria = (cesta > 0).astype(int)

#  Mostrar el resultado final para verificar la estructura
print("--- Tamaño de la matriz binaria ---")
print(cesta_binaria.shape)

print("\n--- Primeras filas de la cesta binaria ---")
print(cesta_binaria.head())
```

```python
# Importar las herramientas necesarias para minería de datos
from mlxtend.frequent_patterns import apriori, association_rules
```

```python

# 1. APLICAR EL ALGORITMO APRIORI
# El algoritmo escanea la 'cesta_binaria' para encontrar los productos que más se compran juntos.
items_frecuentes = apriori(cesta_binaria, min_support=0.03, use_colnames=True)
```

```python
print("--- Combinaciones más frecuentes encontradas ---")
# Ordenamos de mayor a menor soporte para ver los más populares primero
print(items_frecuentes.sort_values(by='support', ascending=False).head())
```

```python
# 2. GENERAR LAS REGLAS DE ASOCIACIÓN
reglas = association_rules(items_frecuentes, metric="lift", min_threshold=1.0)
```

```python
print("\n--- Primeras Reglas de Asociación Generadas ---")
# Imprimimos solo las columnas más importantes para el análisis gerencial
print(reglas[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
```

```python
# MOSTRAR LAS 10 REGLAS MÁS RELEVANTES
# 1. Ordenamos las reglas.
top_10_reglas = reglas.sort_values(by='lift', ascending=False).head(10)

# 2. Filtramos solo las columnas que nos interesan para el análisis gerencial

top_10_limpio = top_10_reglas[['antecedents', 'consequents', 'support', 'confidence', 'lift']]

print("--- Top 10 Reglas de Asociación Más Relevantes ---")
print(top_10_limpio)
```
